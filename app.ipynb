```python
x = torch.tensor(3,)
w=torch.tensor(4., requires_grad=True)
b=torch.tensor(5., requires_grad=True)
```

```python
x,w,b
```

```python
y =w*x+b
y
```

```python
y.backward()
```

```python
print('dy/dx',x.grad)
```

```python
# 1. Create tensors
a = torch.tensor([1, 2, 3, 4])
b = torch.tensor([
    [1, 2],
    [3, 4]
])

# 2. Display properties
print("Tensor A:", a)
print("Tensor B:", b)

print("\nShape of A:", a.shape)
print("Shape of B:", b.shape)

print("\nDimensions of A:", a.ndim)
print("Dimensions of B:", b.ndim)

print("\nData type:", a.dtype)
print("Elements:", b.numel())

# 3. Perform operations
print("\nAddition:", a + 2)
print("Multiplication:", a * 2)

# 4. Reshape
reshaped = a.reshape(2, 2)
print("\nReshaped A:", reshaped)

# 5. Create random tensor
random_tensor = torch.rand(3, 4)
print("\nRandom Tensor:", random_tensor)

# 6. Matrix multiplication
result = torch.matmul(b, b)
print("\nMatrix Multiplication:", result)
```
